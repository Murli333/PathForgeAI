```python
def recommend_jobs(resume , k=5):
    embedding = model.encode(
        [resume],
        convert_to_numpy=True
    ).astype(np.float32)

    faiss.normalize_L2(embedding)
    scores,indices = index.search(
        embedding,
        k
    )

    recommendations = jobs.iloc[indices[0]].copy()
    recommendations["score"] = (scores[0] * 100).round(2)
    return recommendations
```

```python
for category, skills in skills_dict.items():
    patterns = [nlp(skill) for skill in skills]
    matcher.add(category,patterns)
```

```python
def extract_skills(text):
    if pd.isna(text):
        return {}

    text = str(text)
    doc = nlp(text)
    matches = matcher(doc)
    ex1 = {}
    for m_id,s,e in matches:
     category = nlp.vocab.strings[m_id]
     skill = doc[s:e].text
     ex1.setdefault(
        category,
        []
    ).append(skill)
     for key in ex1:
        ex1[key] = sorted(
            list(
                set(ex1[key])
            )
        )
    return ex1
```

```python
from PyPDF2 import PdfReader
```

```python
def extract_text(resume):
    reader = PdfReader(resume)
    text = ""
    for page in reader.pages:
       page_text = page.extract_text()
       if page_text:
         text += page.extract_text() + "\n"
    return text
```

```python
def compare_skills(resume_skills, job_skills):

    analysis = {}

    for category in skills_dict.keys():

        resume_map = {
            skill.lower().strip(): skill
            for skill in resume_skills.get(category, [])
        }

        job_map = {
            skill.lower().strip(): skill
            for skill in job_skills.get(category, [])
        }

        matched_keys = resume_map.keys() & job_map.keys()
        missing_keys = job_map.keys() - resume_map.keys()

        matched = [job_map[k] for k in sorted(matched_keys)]
        missing = [job_map[k] for k in sorted(missing_keys)]

        # keeping things that are useful 
        if matched or missing:
            analysis[category] = {
                "matched": matched,
                "missing": missing
            }

    return analysis
```

```python
def analyze_resume(resume):

    # -------------------------
    # 1. Extract resume text
    # -------------------------
    resume_text = extract_text(resume)

    if not resume_text.strip():
        raise ValueError("Could not extract text from resume.")

    # -------------------------
    # 2. Recommend jobs
    # -------------------------
    recommendations = recommend_jobs(resume_text)

    if recommendations.empty:
        raise ValueError("No relevant jobs found.")

    # -------------------------
    # 3. Best job
    # -------------------------
    top_job = recommendations.iloc[0]

    # -------------------------
    # 4. Job text
    # -------------------------
    job_text = top_job["cleaned"]

    if pd.isna(job_text) or not str(job_text).strip():
        job_text = top_job["description"]

    # -------------------------
    # 5. Extract skills
    # -------------------------
    resume_skills = extract_skills(resume_text)
    job_skills = extract_skills(job_text)

    # Remove empty categories
    resume_skills = {
        category: skills
        for category, skills in resume_skills.items()
        if skills
    }

    job_skills = {
        category: skills
        for category, skills in job_skills.items()
        if skills
    }

    # -------------------------
    # 6. Compare skills
    # -------------------------
    comparison = compare_skills(
        resume_skills,
        job_skills
    )

    # -------------------------
    # 7. Career readiness
    # -------------------------
    matched = sum(
        len(value["matched"])
        for value in comparison.values()
    )

    total = sum(
        len(value["matched"]) +
        len(value["missing"])
        for value in comparison.values()
    )

    skill_readiness = round(
        (matched / total) * 100,
        2
    ) if total else 0

    # -------------------------
    # 8. Strengths
    # -------------------------
    strengths = [
        category
        for category, value in comparison.items()
        if value["matched"]
    ]

    # -------------------------
    # 9. Improvement areas
    # -------------------------
    improvements = [
        category
        for category, value in comparison.items()
        if value["missing"]
    ]

    # -------------------------
    # 10. Learning recommendations
    # -------------------------
    learning = []

    for value in comparison.values():

        for skill in value["missing"]:

            recommendation = roadmap.get(
                skill.lower()
            )

            if recommendation:
                learning.append(recommendation)

    # Remove duplicate recommendations
    learning = list(dict.fromkeys(learning))

    # -------------------------
    # 11. Final report
    # -------------------------
    return {
        "best_fitted_role": top_job["title"],
        "company": top_job["company_name"],
        "location": top_job["location"],
        "similarity_score": round(
            float(top_job["score"]),
            2
        ),
        "career_readiness": skill_readiness,
        "resume_skills": resume_skills,
        "job_skills": job_skills,
        "analysis": comparison,
        "strengths": strengths,
        "improvement_areas": improvements,
        "learning_recommendations": learning,
        "top_recommendations": recommendations
    } 
```
