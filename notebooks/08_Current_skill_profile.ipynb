```python
import re
import time

# Flatten your skill dictionary
flat_skills = []

for category, skills in skills_dict.items():
    for skill in skills:
        flat_skills.append((category, skill))

def extract_skills_fast(text):
    if pd.isna(text):
        return {}

    text = str(text).lower()
    result = {}

    for category, skill in flat_skills:
        pattern = r"\b" + re.escape(skill.lower()) + r"\b"

        if re.search(pattern, text):
            result.setdefault(category, []).append(skill)

    return result
```

```python
sample_resume = """
Computer Science student with experience in Python, C++, JavaScript,
React, SQL, MongoDB, Git, machine learning and TensorFlow.
Built machine learning projects and web applications.
"""

resume_skills = extract_skills_fast(sample_resume)

print(resume_skills)
```

```python
current_skill_profile = set()

for skills in resume_skills.values():
    for skill in skills:
        current_skill_profile.add(skill.lower().strip())

print("Current Skills:", len(current_skill_profile))
print(sorted(current_skill_profile))
```

```python


reliable_role_skill_df = pd.read_csv(
    "../data/req_skill_graph.csv"
)

print(reliable_role_skill_df.shape)
display(reliable_role_skill_df.head())
print(reliable_role_skill_df.columns.tolist())
```

```python
target_role = "Data Architect"
```

```python
target_requirements = reliable_role_skill_df[
    reliable_role_skill_df["role"] == target_role
].copy()

print("Target Role :", target_role)
print("Required skills:", len(target_requirements))

display(
    target_requirements[
        ["skill","skill_demand_pct","requirement_level"]
    ].sort_values("skill_demand_pct",ascending=False)
)
```

```python
current_skills = set(
    skill.lower().strip()
    for skill in current_skill_profile
)

required_skills = set(
    target_requirements["skill"].str.lower().str.strip()
)

matched_skills = current_skills & required_skills
missing_skills = required_skills - current_skills

print("Current skills:", len(current_skills))
print("Required skills:", len(required_skills))
print("Matched skills:", len(matched_skills))
print("Missing skills:", len(missing_skills))

print("\nMATCHED:")
print(sorted(matched_skills))

print("\nMISSING:")
print(sorted(missing_skills))
```

```python
skill_gap_df = target_requirements[
    ~target_requirements["skill"].str.lower().str.strip().isin(current_skills)
].copy()

skill_gap_df = skill_gap_df[
    [
        "skill", "skill_demand_pct" , "requirement_level"
    ]
].sort_values(
    "skill_demand_pct",
    ascending=False
).reset_index(drop=True)

print("Total missing skills :", len(skill_gap_df))

display(skill_gap_df)
```

```python
critical_gaps = skill_gap_df[
    skill_gap_df["requirement_level"] == "Core"
].copy()

print("Critical skill gaps:", len(critical_gaps))

display(critical_gaps)
```

```python
requirement_weights = {
    "Core": 1.00,
    "Common": 0.75,
    "Supporting": 0.50,
    "Rare": 0.25
}

skill_gap_df["requirement_weight"] = (
    skill_gap_df["requirement_level"]
    .map(requirement_weights)
)

skill_gap_df["demand_score"] = (
    skill_gap_df["skill_demand_pct"] / 100
)

skill_gap_df["gap_priority_score"] = (
    skill_gap_df["requirement_weight"]
    * skill_gap_df["demand_score"]
)

skill_gap_df = skill_gap_df.sort_values(
    "gap_priority_score",
    ascending=False
).reset_index(drop=True)

display(skill_gap_df)
```

```python
# Remove previous transferability columns if the cell was partially run
for col in ["role_count", "role_count_x", "role_count_y",
            "transferability_score"]:
    if col in skill_gap_df.columns:
        skill_gap_df.drop(columns=col, inplace=True)


# Count reliable roles requiring each skill
skill_role_counts = (
    reliable_role_skill_df
    .groupby("skill")["role"]
    .nunique()
    .rename("role_count")
)

# Add role count to the gap table
skill_gap_df = skill_gap_df.merge(
    skill_role_counts,
    left_on="skill",
    right_index=True,
    how="left"
)

# Total reliable roles
total_reliable_roles = reliable_role_skill_df["role"].nunique()

# Transferability
skill_gap_df["transferability_score"] = (
    skill_gap_df["role_count"] / total_reliable_roles
)

# Display
display(
    skill_gap_df[
        [
            "skill",
            "role_count",
            "transferability_score",
            "skill_demand_pct",
            "requirement_level",
            "gap_priority_score"
        ]
    ]
)
```
