```python
import re
import time

# Flatten your skill dictionary
flat_skills = []

for category, skills in skills_dict.items():
    for skill in skills:
        flat_skills.append((category, skill))

def extract_skills_fast(text):
    if pd.isna(text):
        return {}

    text = str(text).lower()
    result = {}

    for category, skill in flat_skills:
        pattern = r"\b" + re.escape(skill.lower()) + r"\b"

        if re.search(pattern, text):
            result.setdefault(category, []).append(skill)

    return result
```

```python
job_skill_data = [
    extract_skills_fast(text)
    for text in jobs["description"]
]
```

```python
from collections import defaultdict

skill_job_evidence = defaultdict(list)

for i, skills in enumerate(job_skill_data):

    if not any(skills.values()):
        continue

    description = jobs.iloc[i]["description"]

    if pd.isna(description):
        continue

    description = str(description)

    # Skills detected in this job
    job_skills = set()

    for category_skills in skills.values():
        for skill in category_skills:
            job_skills.add(skill.lower().strip())

    for skill in job_skills:
        skill_job_evidence[skill].append(description)


print("Skills with job evidence:", len(skill_job_evidence))
```

```python
for skill in ["etl", "python", "sql", "communication"]:

    print("\n" + "=" * 70)
    print("SKILL:", skill)
    print("Jobs:", len(skill_job_evidence[skill]))

    print("\nSAMPLE DESCRIPTION:")
    print(skill_job_evidence[skill][0][:1000])
```

```python
import re
import random

def clean_job_description(text):
    text = str(text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove excessive whitespace
    text = re.sub(r'\s+', ' ', text)

    # Remove common job-posting noise
    text = re.sub(
        r'\b(remote|hybrid|onsite|full[- ]time|part[- ]time)\b',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Clean whitespace again
    text = re.sub(r'\s+', ' ', text).strip()

    return text

def sample_skill_evidence(skill,n=10):
    descriptions = skill_job_evidence.get(skill,[])

    if not descriptions:
        return[]

    # remove duplicates
    descriptions = list(set(descriptions))

    # dont req more than available
    n = min(n,len(descriptions))

    return random.sample(descriptions,n)
```

```python
for skill in ["etl", "python", "sql"]:

    print("\n" + "=" * 70)
    print("SKILL:", skill)

    samples = sample_skill_evidence(skill, n=3)

    for i, text in enumerate(samples, 1):
        cleaned = clean_job_description(text)

        print(f"\n--- JOB {i} ---")
        print(cleaned[:1000])
```

```python
AI_EVIDENCE_LIMIT = 10

ai_evidence = []

for skill in skill_job_evidence:
    samples = sample_skill_evidence(
        skill,
        n=AI_EVIDENCE_LIMIT
    )

    for text in samples:
        cleaned = clean_job_description(text)
        if len(cleaned) >= 100:
            ai_evidence.append({
                "skill" : skill,
                "description" : cleaned
            })

ai_evidence_df = pd.DataFrame(ai_evidence)
print("Total evidence records :" , len(ai_evidence_df))
print("Unique skills:",ai_evidence_df["skill"].nunique())
ai_evidence_df.to_csv('ai_evidence_df',index=False)
```

```python
def build_ai_exposure_prompt(skill, descriptions):

    evidence = "\n\n".join(
        f"JOB {i+1}:\n{desc[:2000]}"
        for i, desc in enumerate(descriptions)
    )

    prompt = f"""
You are analyzing career resilience and AI exposure for PathForge.

SKILL:
{skill}

JOB POSTING EVIDENCE:
{evidence}

Analyze the work associated with this skill based ONLY on the provided
job-posting evidence and your general understanding of how AI affects work.

Identify the recurring tasks associated with this skill and determine
whether those tasks are primarily:

1. AI-substitutable
   - AI can potentially perform a large portion of the task with limited
     human intervention.

2. AI-augmented
   - AI can significantly assist the human, but human expertise,
     verification, judgment, or decision-making remains important.

3. AI-complementary
   - The task primarily depends on human judgment, collaboration,
     leadership, domain context, physical interaction, or other capabilities
     that AI does not directly replace.

Return ONLY valid JSON in this structure:

{{
    "skill": "{skill}",
    "tasks": [
        {{
            "task": "task description",
            "classification": "AI-substitutable | AI-augmented | AI-complementary",
            "reason": "brief explanation"
        }}
    ],
    "overall_classification": "AI-substitutable | AI-augmented | AI-complementary",
    "ai_exposure_score": 0.0,
    "summary": "brief overall explanation"
}}

The ai_exposure_score must be between 0 and 1,
where 0 means very low AI exposure and 1 means very high AI exposure.
"""

    return prompt
```
