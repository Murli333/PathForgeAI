````python
import re
import time

# Flatten your skill dictionary
flat_skills = []

for category, skills in skills_dict.items():
    for skill in skills:
        flat_skills.append((category, skill))

def extract_skills_fast(text):
    if pd.isna(text):
        return {}

    text = str(text).lower()
    result = {}

    for category, skill in flat_skills:
        pattern = r"\b" + re.escape(skill.lower()) + r"\b"

        if re.search(pattern, text):
            result.setdefault(category, []).append(skill)

    return result
````

````python
job_skill_data = [
    extract_skills_fast(text)
    for text in jobs["description"]
]
````

````python
from collections import defaultdict

skill_job_evidence = defaultdict(list)

for i, skills in enumerate(job_skill_data):

    if not any(skills.values()):
        continue

    description = jobs.iloc[i]["cleaned"]

    if pd.isna(description):
        continue

    description = str(description)

    # Skills detected in this job
    job_skills = set()

    for category_skills in skills.values():
        for skill in category_skills:
            job_skills.add(skill.lower().strip())

    for skill in job_skills:
        skill_job_evidence[skill].append(description)


print("Skills with job evidence:", len(skill_job_evidence))
````

````python
import re
import random

def clean_job_description(text):
    text = str(text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove excessive whitespace
    text = re.sub(r'\s+', ' ', text)

    # Remove common job-posting noise
    text = re.sub(
        r'\b(remote|hybrid|onsite|full[- ]time|part[- ]time)\b',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # Clean whitespace again
    text = re.sub(r'\s+', ' ', text).strip()

    return text

def sample_skill_evidence(skill,n=10):
    descriptions = skill_job_evidence.get(skill,[])

    if not descriptions:
        return[]

    # remove duplicates
    descriptions = list(set(descriptions))

    # dont req more than available
    n = min(n,len(descriptions))

    return random.sample(descriptions,n)
````

````python
AI_EVIDENCE_LIMIT = 10

ai_evidence = []

for skill in skill_job_evidence:
    samples = sample_skill_evidence(
        skill,
        n=AI_EVIDENCE_LIMIT
    )

    for text in samples:
        cleaned = clean_job_description(text)
        if len(cleaned) >= 100:
            ai_evidence.append({
                "skill" : skill,
                "description" : cleaned
            })

ai_evidence_df = pd.DataFrame(ai_evidence)
print("Total evidence records :" , len(ai_evidence_df))
print("Unique skills:",ai_evidence_df["skill"].nunique())
ai_evidence_df.to_csv('ai_evidence_df',index=False)
````

````python
def build_ai_exposure_prompt(skill, descriptions):

    evidence = "\n\n".join(
        f"JOB {i+1}:\n{desc[:2000]}"
        for i, desc in enumerate(descriptions)
    )

    prompt = f"""
You are analyzing career resilience and AI exposure for PathForge.

SKILL:
{skill}

JOB POSTING EVIDENCE:
{evidence}

Analyze the work associated with this skill based ONLY on the provided
job-posting evidence and your general understanding of how AI affects work.

Identify the recurring tasks associated with this skill and determine
whether those tasks are primarily:

1. AI-substitutable
   - AI can potentially perform a large portion of the task with limited
     human intervention.

2. AI-augmented
   - AI can significantly assist the human, but human expertise,
     verification, judgment, or decision-making remains important.

3. AI-complementary
   - The task primarily depends on human judgment, collaboration,
     leadership, domain context, physical interaction, or other capabilities
     that AI does not directly replace.

Return ONLY valid JSON in this structure:

{{
    "skill": "{skill}",
    "tasks": [
        {{
            "task": "task description",
            "classification": "AI-substitutable | AI-augmented | AI-complementary",
            "reason": "brief explanation"
        }}
    ],
    "overall_classification": "AI-substitutable | AI-augmented | AI-complementary",
    "ai_exposure_score": 0.0,
    "summary": "brief overall explanation"
}}

The ai_exposure_score must be between 0 and 1,
where 0 means very low AI exposure and 1 means very high AI exposure.
"""

    return prompt
````

````python
import json
import time
import re
import pandas as pd
from pathlib import Path

# ---------- CONFIGURATION ----------
MODEL_NAME = "gemini-3.1-flash-lite"
BATCH_SIZE = 20
EVIDENCE_PER_SKILL = 5
MAX_RETRIES = 3

DATA_DIR = Path("../data")
DATA_DIR.mkdir(parents=True, exist_ok=True)

CHECKPOINT_PATH = DATA_DIR / "ai_exposure_checkpoint.json"
OUTPUT_PATH = DATA_DIR / "ai_exposure_results.csv"


# ---------- ROBUST JSON PARSER ----------
def parse_ai_exposure_response(response_text):
    if not response_text:
        raise ValueError("Gemini returned an empty response.")

    text = response_text.strip()

    # Handle JSON wrapped in Markdown code fences
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text, count=1, flags=re.I)
        text = re.sub(r"\s*```$", "", text, count=1)

    # Extract the first valid JSON object
    start = text.find("{")
    if start == -1:
        raise ValueError("No JSON object found in the response.")

    result, _ = json.JSONDecoder().raw_decode(text[start:])

    required_fields = {
        "skill",
        "tasks",
        "overall_classification",
        "ai_exposure_score",
        "summary"
    }

    if not isinstance(result, dict):
        raise ValueError("Response must be a JSON object.")

    if not required_fields.issubset(result.keys()):
        raise ValueError("Response is missing required fields.")

    valid_classes = {
        "AI-substitutable",
        "AI-augmented",
        "AI-complementary"
    }

    if result["overall_classification"] not in valid_classes:
        raise ValueError("Invalid overall classification.")

    score = float(result["ai_exposure_score"])

    if not 0 <= score <= 1:
        raise ValueError("AI exposure score must be between 0 and 1.")

    if not isinstance(result["tasks"], list):
        raise ValueError("Tasks must be a list.")

    result["ai_exposure_score"] = score
    return result


# ---------- LOAD CHECKPOINT ----------
if CHECKPOINT_PATH.exists():
    with open(CHECKPOINT_PATH, "r", encoding="utf-8") as file:
        checkpoint = json.load(file)
else:
    checkpoint = {"results": {}, "errors": {}}

checkpoint.setdefault("results", {})
checkpoint.setdefault("errors", {})

completed_skills = set(checkpoint["results"].keys())

remaining_skills = [
    skill for skill in skill_job_evidence
    if skill not in completed_skills
]

batch_skills = remaining_skills[:BATCH_SIZE]

print(f"Total skills: {len(skill_job_evidence)}")
print(f"Already completed: {len(completed_skills)}")
print(f"Remaining: {len(remaining_skills)}")
print(f"Skills in this batch: {len(batch_skills)}")


# ---------- SAVE CHECKPOINT + CSV ----------
def save_ai_exposure_outputs():
    temp_path = CHECKPOINT_PATH.with_suffix(".tmp")

    with open(temp_path, "w", encoding="utf-8") as file:
        json.dump(checkpoint, file, ensure_ascii=False, indent=2)

    temp_path.replace(CHECKPOINT_PATH)

    rows = []

    for skill, result in checkpoint["results"].items():
        rows.append({
            "skill": skill,
            "overall_classification": result["overall_classification"],
            "ai_exposure_score": result["ai_exposure_score"],
            "summary": result["summary"],
            "tasks": json.dumps(
                result["tasks"],
                ensure_ascii=False
            )
        })

    pd.DataFrame(
        rows,
        columns=[
            "skill",
            "overall_classification",
            "ai_exposure_score",
            "summary",
            "tasks"
        ]
    ).to_csv(OUTPUT_PATH, index=False)


# ---------- PROCESS CURRENT BATCH ----------
for position, skill in enumerate(batch_skills, start=1):

    print(f"\n[{position}/{len(batch_skills)}] Analyzing: {skill}")

    descriptions = sample_skill_evidence(
        skill,
        n=EVIDENCE_PER_SKILL
    )

    if not descriptions:
        checkpoint["errors"][skill] = "No job evidence available."
        save_ai_exposure_outputs()
        continue

    success = False

    for attempt in range(MAX_RETRIES + 1):
        try:
            prompt = build_ai_exposure_prompt(skill, descriptions)

            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt
            )

            result = parse_ai_exposure_response(response.text)

            # Ensure the result corresponds to the requested skill
            result["skill"] = skill

            checkpoint["results"][skill] = result
            checkpoint["errors"].pop(skill, None)

            save_ai_exposure_outputs()

            print(
                f"  Classification: {result['overall_classification']}\n"
                f"  Exposure score: {result['ai_exposure_score']:.2f}\n"
                f"  Tasks identified: {len(result['tasks'])}\n"
                f"  Saved successfully."
            )

            success = True
            break

        except Exception as error:
            if attempt < MAX_RETRIES:
                wait_seconds = 3 * (2 ** attempt)

                print(
                    f"  Attempt {attempt + 1} failed: {error}\n"
                    f"  Retrying in {wait_seconds} seconds..."
                )

                time.sleep(wait_seconds)

            else:
                checkpoint["errors"][skill] = str(error)

                print(f"  Failed after retries: {error}")
                save_ai_exposure_outputs()

    # Small pause between skills to reduce request pressure
    time.sleep(2)


# ---------- BATCH SUMMARY ----------
save_ai_exposure_outputs()

print("\nBATCH COMPLETE")
print(f"Successfully analyzed: {len(checkpoint['results'])}")
print(f"Recorded errors: {len(checkpoint['errors'])}")
print(f"Checkpoint: {CHECKPOINT_PATH}")
print(f"CSV output: {OUTPUT_PATH}")
````
