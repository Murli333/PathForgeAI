```python
import re
import time

# Flatten your skill dictionary
flat_skills = []

for category, skills in skills_dict.items():
    for skill in skills:
        flat_skills.append((category, skill))

def extract_skills_fast(text):
    if pd.isna(text):
        return {}

    text = str(text).lower()
    result = {}

    for category, skill in flat_skills:
        pattern = r"\b" + re.escape(skill.lower()) + r"\b"

        if re.search(pattern, text):
            result.setdefault(category, []).append(skill)

    return result
```

```python
job_skill_data = [
    extract_skills_fast(text)
    for text in jobs["description"]
]
```

```python
from collections import defaultdict

skill_job_evidence = defaultdict(list)

for i, skills in enumerate(job_skill_data):

    if not any(skills.values()):
        continue

    description = jobs.iloc[i]["description"]

    if pd.isna(description):
        continue

    description = str(description)

    # Skills detected in this job
    job_skills = set()

    for category_skills in skills.values():
        for skill in category_skills:
            job_skills.add(skill.lower().strip())

    for skill in job_skills:
        skill_job_evidence[skill].append(description)


print("Skills with job evidence:", len(skill_job_evidence))
```

```python
for skill in ["etl", "python", "sql", "communication"]:

    print("\n" + "=" * 70)
    print("SKILL:", skill)
    print("Jobs:", len(skill_job_evidence[skill]))

    print("\nSAMPLE DESCRIPTION:")
    print(skill_job_evidence[skill][0][:1000])
```
