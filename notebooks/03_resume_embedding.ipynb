```python
import numpy as np 
import pandas as pd
import faiss
from sentence_transformers import SentenceTransformer
```

```python
jobs = pd.read_csv(r"C:\Users\murli mishra\OneDrive\Desktop\PYTHON\PathForgeAI\data\jobs_tech_master.csv")
```

```python
jobs.head()
```

```python
jobs['cleaned'] = jobs['cleaned'].fillna('')
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
embdeddings = model.encode(
    jobs['cleaned'].tolist(),
    batch_size=64,
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
embdeddings[0][:4]
```

```python
embdeddings = embdeddings.astype(np.float32)
```

```python
dimension = embdeddings.shape[1]

index = faiss.IndexFlatIP(
    dimension
)
```

```python
faiss.normalize_L2(embdeddings)
```

```python
index.add(
    embdeddings
)
```

```python
faiss.write_index(
    index,
    "PathForge.index"
)
```
