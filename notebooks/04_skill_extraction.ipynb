```python
matcher = PhraseMatcher(
    nlp.vocab,
    attr = "LOWER"
)
```

```python
for category ,skills in skill_dict.items():
    patterns = [
        nlp.make_doc(skill)
        for skill in skills
    ]
    matcher.add(
        category,
        patterns
    )
```

```python
resume = """

Python Developer

TensorFlow

PyTorch

AWS

Docker

Git

Linux

MySQL

"""
```

```python
doc = nlp(resume)
```

```python
matches = matcher(doc)
matches
```

```python
for match_id,start,end in matches:
    category = nlp.vocab.strings[
        match_id
    ]
    skill = doc[start:end].text
    print(category, ":" , skill)
```

```python
extracted = {}
for match_id,start,end in matches:
    category = nlp.vocab.strings[match_id]
    skill = doc[start:end].text

    extracted.setdefault(
        category,
        []
    ).append(skill)
```

```python
extracted
```

```python
for key in extracted:
    extracted[key] = sorted(
        list(
            set(extracted[key])
        )
    )
```
