```python
import faiss
import numpy as np 
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

```python
jobs = [
    "Machine Learning Engineer",
    "Python Backend Developer",
    "Frontend React Developer",
    "Data Scientist",
    "Software Engineer",
    "Cloud Engineer",
    "DevOps Engineer",
    "AI Research Engineer"
    "Machine Learning Engineer with Python, TensorFlow and AWS",
    "Frontend React Developer with TypeScript",
    "Data Scientist using SQL and Power BI",
    "DevOps Engineer using Docker and Kubernetes"
]
```

```python
job_vectors = model.encode(jobs)
```

```python
job_vectors.shape
```

```python
job_vectors = np.array(
    job_vectors,
    dtype=np.float32
)
```

```python
dim = job_vectors.shape[1]
index = faiss.IndexFlatL2(
    dim

)
```

```python
index.add(job_vectors)
```

```python
index.ntotal
```

```python
resume = [ "Experienced Machine Learning Engineer skilled in Python and TensorFlow."
]
resume_vector = model.encode(resume)
resume_vector = np.array(
    resume_vector,
    dtype=np.float32
)
```

```python
k = 3
dist,indices = index.search(
    resume_vector,
    k
)
```

```python
for i in indices[0]:
    print(jobs[i])
```

```python
for d, i in zip(
    dist[0],
    indices[0]
):

    print(
        jobs[i],
        d
    )
```

```python
faiss.write_index(index,"PathForge.index")
```

```python
loaded_index = faiss.read_index('PathForge.index')
```

```python
d,ind = loaded_index.search(resume_vector,2)
for d,i in zip(dist[0],ind[0]):
    print(f'{jobs[i]} : {d}')
```
