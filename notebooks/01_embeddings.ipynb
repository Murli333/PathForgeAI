```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer(
    'all-MiniLM-L6-v2'
)
```

```python
jobs = [
    "Machine Learning Engineer",
    "Python Backend Developer",
    "Frontend React Developer",
    "Data Scientist",
    "Software Engineer"
]
```

```python
job_vectors = model.encode(jobs)
```

```python
job_vectors.shape
```

```python
job_vectors[0][:20]
```

```python
resume = [
    "Frontend developer skilled in React and JavaScript."
]
resume_vector = model.encode(resume)
```

```python
resume_vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
scores = cosine_similarity(
    resume_vector,job_vectors
)
scores.shape
```

```python
for job, score in zip(jobs,scores[0]):
    print(f'{job} : {score: .4f}')
```
