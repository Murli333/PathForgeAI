```python
def extract_skills(text):
    if pd.isna(text):
        return {}

    text = str(text)
    doc = nlp(text)
    matches = matcher(doc)
    ex1 = {}
    for m_id,s,e in matches:
     category = nlp.vocab.strings[m_id]
     skill = doc[s:e].text
     ex1.setdefault(
        category,
        []
    ).append(skill)
     for key in ex1:
        ex1[key] = sorted(
            list(
                set(ex1[key])
            )
        )
    return ex1
```

```python
def extract_skills_batch(texts, batch_size=64):

    results = []

    for doc in nlp.pipe(texts, batch_size=batch_size):

        matches = matcher(doc)

        extracted = {}

        for m_id, s, e in matches:

            category = nlp.vocab.strings[m_id]
            skill = doc[s:e].text

            extracted.setdefault(category, []).append(skill)

        for category in extracted:
            extracted[category] = sorted(set(extracted[category]))

        results.append(extracted)

    return results
```

```python
import time

start = time.time()

test_skills = extract_skills_batch(
    jobs["description"].head(100)
)

elapsed = time.time() - start

print("Processed:", len(test_skills))
print("Time:", round(elapsed, 2), "seconds")
```

```python
import re
import time

# Flatten your skill dictionary
flat_skills = []

for category, skills in skills_dict.items():
    for skill in skills:
        flat_skills.append((category, skill))

def extract_skills_fast(text):
    if pd.isna(text):
        return {}

    text = str(text).lower()
    result = {}

    for category, skill in flat_skills:
        pattern = r"\b" + re.escape(skill.lower()) + r"\b"

        if re.search(pattern, text):
            result.setdefault(category, []).append(skill)

    return result
```

```python
start = time.time()

fast_test = [
    extract_skills_fast(text)
    for text in jobs["description"].head(500)
]

elapsed = time.time() - start

print("Processed:", len(fast_test))
print("Time:", round(elapsed, 2), "seconds")
```

```python
for i in range(10):
    slow = extract_skills(jobs.iloc[i]["description"])
    fast = extract_skills_fast(jobs.iloc[i]["description"])

    print("=" * 70)
    print("JOB:", jobs.iloc[i]["title"])
    print("SLOW:", slow)
    print("FAST:", fast)
```

```python
def flatten_skills(skill_result):
    return {
        skill.lower()
        for skills in skill_result.values()
        for skill in skills
    }


differences = []

for i in range(100):
    slow = flatten_skills(
        extract_skills(jobs.iloc[i]["description"])
    )

    fast = flatten_skills(
        extract_skills_fast(jobs.iloc[i]["description"])
    )

    differences.append({
        "job": jobs.iloc[i]["title"],
        "slow_count": len(slow),
        "fast_count": len(fast),
        "extra_fast": len(fast - slow),
        "missing_fast": len(slow - fast)
    })

comparison_df = pd.DataFrame(differences)

print("Average slow skills:",
      round(comparison_df["slow_count"].mean(), 2))

print("Average fast skills:",
      round(comparison_df["fast_count"].mean(), 2))

print("Average extra fast:",
      round(comparison_df["extra_fast"].mean(), 2))

print("Average missing fast:",
      round(comparison_df["missing_fast"].mean(), 2))
```
